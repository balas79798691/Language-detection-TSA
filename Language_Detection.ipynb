# 🌍 Language Detection

### Objective
Detect the language of a user-provided sentence using the `langdetect` library.

**Concept:** Natural Language Processing (NLP)


```python
!pip install langdetect
```

```python
from langdetect import detect, detect_langs

text = input("Enter a sentence: ")

if not text.strip():
    print("Please enter some text.")
else:
    language = detect(text)
    probabilities = detect_langs(text)

    language_names = {
        "en": "English", "ta": "Tamil", "hi": "Hindi",
        "te": "Telugu", "ml": "Malayalam", "kn": "Kannada",
        "fr": "French", "de": "German", "es": "Spanish",
        "it": "Italian", "pt": "Portuguese", "ja": "Japanese",
        "ko": "Korean", "zh-cn": "Chinese", "ar": "Arabic",
        "ru": "Russian"
    }

    print("\n===== LANGUAGE DETECTION =====")
    print("Input:", text)
    print("Detected Language:", language_names.get(language, language))
    print("Language Code:", language)
    print("Probabilities:", probabilities)
```

## Example

Try:

`Hello, how are you today?`

or

`வணக்கம், நீங்கள் எப்படி இருக்கிறீர்கள்?`

The program identifies the language and displays its language code.


## Conclusion

The program successfully detects the language of a given sentence using NLP-based language detection.
